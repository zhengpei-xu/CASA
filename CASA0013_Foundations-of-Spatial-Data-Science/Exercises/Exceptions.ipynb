### Unit Tests

```python
import unittest

def fun(x):
  return x + 1

class MyTest(unittest.TestCase):
  def test(self):
    self.assertEqual(fun(3), 4)
    print("Assertion 1 passed.")
    self.assertEqual(fun(3), 5)
    print("Assertion 2 passed.")

m = MyTest()
m.test()
```

### docstring test

```python
def square(x):
    """Return the square of x.

    >>> square(2)
    4
    >>> square(-2)
    4
    >>> square(-1)
    2
    """
    return x * x
    
# 控制代码的执行时机和启动 doctest 模块进行自我测试
if __name__ == '__main__':
    import doctest
    doctest.testmod()
```
