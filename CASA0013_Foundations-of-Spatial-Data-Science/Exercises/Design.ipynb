# Design in python programming

## Procedural programming

易读、流程清晰，就像一个分步骤的指南。数据在函数之间流动，但数据和函数本身是分离的

```python
# 数据
scores = [65, 78, 92, 45, 88]

# 步骤 1：定义一个函数来增加分数
def add_bonus(original_scores):
    """过程：按顺序遍历列表并修改/生成新列表。"""
    new_scores = []
    for score in original_scores:
        # 这是一个指令
        bonus_score = score + 10 
        new_scores.append(bonus_score)
    return new_scores

# 步骤 2：定义一个函数来筛选
def filter_passing(scores_list, threshold):
    """过程：按顺序遍历列表并检查条件。"""
    passing_scores = []
    for score in scores_list:
        if score > threshold: # 又一个指令
            passing_scores.append(score)
    return passing_scores

# 主流程：按顺序执行步骤
processed_scores = add_bonus(scores)
final_results = filter_passing(processed_scores, 80)

print(f"原始分数: {scores}")
print(f"加分后: {processed_scores}")
print(f"通过筛选: {final_results}")
```

## Functional programming

没有 for 循环！代码更关注于数据应该如何转换（map 和 filter），而不是如何一步步执行。每个函数都是独立的“转换站”，不改变原始数据，非常优雅

```python
from functools import reduce

# 数据（不可变）
scores = [65, 78, 92, 45, 88]

# 1. 定义纯函数 (Pure Functions) - 只依赖输入，不改变外部状态
# lambda 是定义小型匿名函数的完美选择
add_10 = lambda x: x + 10
is_passing = lambda x: x > 80

# 2. 数据转换：使用 map() (应用函数到列表的每个元素)
# 这一步是“数据流”的第一站
scores_plus_10 = list(map(add_10, scores)) 

# 3. 数据过滤：使用 filter() (根据条件函数筛选列表元素)
final_results_fp = list(filter(is_passing, scores_plus_10))

print(f"原始分数: {scores}")
print(f"加分后 (Map): {scores_plus_10}")
print(f"通过筛选 (Filter): {final_results_fp}")
```

## Object oriented programming

所有的分数和计算都紧密地捆绑在 Student 对象的内部。如果你要管理数千个学生，OOP能让你用更直观、更结构化的方式来操作和维护。

```python
class Student:
    """这是一个蓝图（Class），定义了学生对象的属性和行为。"""
    
    def __init__(self, name, score):
        # 属性（数据）：封装在对象内部
        self.name = name
        self.score = score
        self.bonus_score = 0
    
    # 方法（行为）：处理对象内部的数据
    def apply_bonus(self, bonus_amount=10):
        """计算加分后的成绩，并将结果存储在对象的属性中。"""
        self.bonus_score = self.score + bonus_amount
        print(f"{self.name} 的分数从 {self.score} 更新为 {self.bonus_score}")
    
    # 方法：检查对象是否通过
    def check_passing(self, threshold=80):
        """检查学生是否达到门槛。"""
        return self.bonus_score > threshold

# 实例化对象（Objects）：根据蓝图创建具体的实体
students = [
    Student("Alice", 65),
    Student("Bob", 92),
    Student("Charlie", 78)
]

print("--- 执行操作 ---")
# 流程：与对象互动
passing_students = []
for student in students:
    student.apply_bonus() # 调用对象的方法
    
    if student.check_passing():
        passing_students.append(student.name)

print("--- 结果 ---")
print(f"原始学生数量: {len(students)}")
print(f"通过的学生: {passing_students}")
```
