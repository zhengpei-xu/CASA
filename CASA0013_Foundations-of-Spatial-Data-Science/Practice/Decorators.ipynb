```python
def better(func):
    return func().title() + "!"

@better
def hello():
    return "hello world"

hello()
```

```python
def hello():
    return "hello world"
    
def better(func):
    return func().title() + "!"

better(hello)
```

```python
from functools import wraps

def simple_logger(func):
    @wraps(func)
    def wrapper(*args, **kwargs):
        print(f"executing the function: {func.__name__} with arguments {args}")
        print(f"the result is {func(*args)}")
        pass
    return wrapper

@simple_logger
def add(a, b):
    return a+b
```
