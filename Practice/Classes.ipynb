Cube: inheritance from shape

```python
class cube(shape):
    def __init__(self, e):
        self.shape_type = "Cube"
        self.edge = e
        
    def volume(self):
        return self.edge**3
    def diameter(self):
        return (self.edge**2 + self.edge**2)**(1/2)
```
